# How does a nineteenth-century tactile Bible challenge what we mean by accessibility in the digital age?

### Rethinking touch, reading, and access through a raised-type New Testament

## A book that asks to be touched

Many cultural heritage objects reach us through glass cases, catalogue records, or digital images. These formats make objects visible, but they also shape how we are allowed to encounter them.

This project begins with a book that challenges a familiar habit: the idea that access mainly means being able to see, read, or search for information.

The object at the centre of this notebook is a nineteenth-century raised-type New Testament designed for blind readers. Its letters were not only printed to be seen. They were formed as raised shapes so that the text could be followed by touch.

This makes the book more than a religious text in an unusual format. It is also a reading technology. It invites us to ask a wider question:

> **What does accessibility mean when a book designed for touch is encountered mainly through a screen?**

This question matters because digital access is often treated as the solution to cultural heritage access. Digitisation can make objects easier to find, view, study, and share. But for a tactile book, access cannot be measured only by visibility or information (Green, 2018). The way the object was meant to be read, through fingers, movement, pressure, and pace, also matters.

By placing this raised-type Bible beside today’s ideas of digital accessibility, this project explores what digitisation makes possible, what it cannot fully preserve, and how generative AI can help imagine more public-facing ways of communicating tactile access.

The central argument of the notebook is that accessibility is not only about making information available. For tactile objects, access also means understanding how they were meant to be used, felt, and read.

## Meet the object: a raised-type New Testament

The object at the centre of this project is an English New Testament printed in 1850 and held by the University of Glasgow Special Collections.

It was produced in raised type, also described as high-relief print. This means that the letters stand up from the page as tactile forms. The book was designed so that blind readers could access the text through their fingers rather than through ordinary visual print.

This notebook therefore looks at the Bible not only as scripture, but also as a technology for reading by touch. It belongs to a longer history of attempts to make texts available to readers who were excluded by standard visual print.

## Reading through touch

Tactile reading is not simply visual reading with the fingers replacing the eyes. It has its own rhythm, limits, and skills.

A sighted reader can often scan a whole page, move quickly between lines, or step back to see layout and structure. A tactile reader meets the page through contact. The fingers move across the surface, letter by letter and line by line. Meaning is built through pressure, movement, direction, memory, and repetition.

This makes reading slower and more physical. The reader has to follow the direction of the lines, recognise the shape of each letter, remember what has already been felt, and build meaning step by step.

For this Bible, the raised letters are not just a visible feature of the page. They are the path through which the text is reached.

> **The page is not only a container for words. It is the surface that makes reading possible.**

For a modern viewer looking at a digital image of the Bible, this is easy to miss. A screen can show the raised letters, but it cannot reproduce the feeling of following them by touch. It can show the surface of the page, but not the pressure, movement, and bodily attention involved in tactile reading.

## Accessibility before the digital age

Accessibility did not begin with digital technology.

Today, accessibility is often discussed through tools such as screen readers, captions, alt text, zoom functions, searchable text, audio description, and adjustable interfaces. These tools are important. They can reduce barriers for people who cannot easily use standard visual formats.

But the raised-type New Testament shows that accessibility also has a material history.

Before online catalogues, screen readers, and searchable text, access to reading was also explored through paper, print, raised surfaces, and letterforms. In this book, accessibility was not separate from the object’s design. The shape, height, spacing, and surface of the letters all affected how the text could be read.

The book also belongs to a nineteenth-century history of embossed reading systems for blind readers. It is not the dot-based Braille system that is most familiar today. Instead, it uses raised Roman letters: alphabetic forms that keep the shape of ordinary printed letters, but lift them from the page so they can be traced by touch (Warne, 2025).

<img src="media/raised-type-bible-detail.jpg" alt="Close-up of the raised Roman letters" width="500"/>

*Figure 2. Close-up of the raised Roman letters. Photograph by the author.*

This matters because raised Roman type reflects a particular idea of access. It tries to make printed letters readable through touch while still keeping a close relationship with the visual alphabet (Tilley, 2018; Warne, 2025). This could have made the letters more recognisable to sighted teachers, printers, and institutions. But it also reveals a tension: a form that is familiar to the eye is not always the easiest form for the hand (Tilley, 2018; Warne, 2025).

The raised-type Bible therefore helps us see accessibility not as a single fixed solution, but as something shaped by technology, culture, institutions, and bodies.

## When tactile access becomes digital

Digitisation can make this raised-type Bible easier to encounter.

A digital image can show its pages to people who may never visit the archive. A catalogue record can preserve important details such as date, format, language, and collection location. A viewer can zoom in, compare the raised letters, and return to the object online.

These forms of access matter. They make the Bible more visible, searchable, and shareable. They can also protect the fragile original by reducing the need for physical handling.

But digitisation also changes the form of access.

In the nineteenth century, the raised-type Bible translated printed text into touch. Today, a digital image translates that tactile object back into something mainly visual. The book becomes easier to find and view, but the reading experience it was designed for becomes harder to understand.

This is the central tension of the project:

> **Digitisation can widen access to the Bible as a cultural heritage object, while narrowing access to it as a tactile reading experience.**

This does not mean that digitisation fails. It means that digitisation translates. It changes how an object is encountered.

For a touch-based book, the key question is therefore not only:

> **Is the Bible accessible online?**

It is also:

> **How could its tactile form be made meaningful to people who cannot touch the original?**

| Digital access can make available | But it may not fully preserve |
|---|---|
| A clear image of the raised letters | The feeling of their height, texture, and resistance |
| Catalogue information | The bodily process of tactile reading |
| Remote viewing | Physical contact with the page |
| Searchable or transcribed text | The raised Roman letters as touchable forms |
| Easier sharing and teaching | The slow pace of reading by hand |

This tension is where the creative part of the project begins.

## From question to prototype: making touch interpretable

The previous section showed the main problem this project responds to: a digital image can show the raised letters of the Bible, but it cannot let visitors feel how the book was meant to be read.

The creative task was to imagine a way for visitors to understand touch without touching the fragile original book.

My response is a **Dynamic Tactile Reading Station**: a speculative museum-style prototype inspired by the raised-type New Testament.

The station would keep the original book protected, while offering a separate tactile surface where visitors could explore the idea of raised Roman letters through touch, audio, and guided interaction. It does not try to recreate the exact experience of nineteenth-century blind readers. Instead, it asks how tactile reading might be explained and partly reintroduced when direct handling is not possible.

AI helped me test visual directions, interaction flows, and public-facing explanations. The historical interpretation and final decisions remained human-directed.

### Prompt development and human editing

Generative AI was useful because it could quickly produce possible visual and interaction directions. However, the first outputs were not automatically suitable. Some images made the station look like an ordinary touchscreen kiosk. Others confused raised Roman letters with Braille, made the tactile panel look static, or included invented museum labels.

In this project, the AI tool was used through a prompt-and-output process. The input was a written prompt describing the object, the intended audience, the museum setting, the visual style, and the interaction sequence. The output was a set of generated images or text suggestions that could be reviewed and edited. These outputs were shaped by patterns in the model’s training data, which means they were useful for generating possibilities, but not reliable as historical evidence. For this reason, I treated the AI outputs as draft material. Each output had to be checked against the object, the project aim, and the need to avoid misleading details, such as confusing raised Roman letters with Braille.

The workflow was therefore not a simple process of asking AI for a finished design. It was an iterative cycle of prompting, evaluating, correcting, and re-prompting. Each round helped me make the idea more specific: the protected original Bible, the separate tactile panel, the raised Roman letters, the pin-matrix mechanism, and the guided audio sequence.

## The Dynamic Tactile Reading Station

The proposed station is designed for a museum or digital heritage setting where the original raised-type Bible cannot be handled directly.

The station keeps the original raised-type Bible protected behind glass. Beside it, a separate tactile panel gives visitors a way to explore raised Roman letters without touching the original book.

The panel is imagined as a surface made from many small movable pins. These pins rise to form one word at a time. A visitor follows the raised word by touch, while a small display and audio guide support the reading sequence.

The station brings together three forms of access:

| Form of access | How it works in the prototype |
|---|---|

The original book remains protected, but the idea of tactile reading becomes something visitors can physically explore.

## How the station works

The station is designed to make tactile reading easier to understand by slowing the reading process down. Instead of showing a whole page at once, it presents a short passage one word at a time. This gives visitors time to notice the shape of each raised word, the movement of their fingers, and the rhythm of reading by touch.

<img src="media/interaction-flow-diagram.jpg" alt="AI-generated interaction flow of the Dynamic Tactile Reading Station" width="600"/>

*Figure 5. AI-generated interaction flow of the Dynamic Tactile Reading Station. This is a speculative process diagram, not a working technical system.*

The interaction begins when a visitor selects a short passage on the same reading station. The selected passage is then divided into individual words. On the tactile panel, small pins rise from the flat surface to form the current word in raised Roman letters. The visitor follows the raised letters by touch. When the word is complete, the audio confirms it, and the panel changes to the next word.

| Step | Visitor experience |
|---|---|
| 1 | Select a short passage on the reading station |
| 2 | The passage is divided into individual words |
| 3 | Pins rise to form the current word |
| 4 | The visitor follows the raised Roman letters by touch |
| 5 | Audio confirms the word |
| 6 | The panel changes to the next word |

This sequence connects touch, sound, memory, and pace. The visitor is not only looking at a digital image of raised text. They are invited to experience reading as a slower physical process: feeling a word, hearing it confirmed, and then moving on.

In a real version of the station, this word-by-word sequence would need to be prepared before the tactile panel changes from one word to the next. The small code example below shows how that sequence could be organised.

### Notebook demonstration: preparing a word-by-word reading sequence

The following small code example shows one simple step that a real version of the station would need: preparing a selected passage as a word-by-word reading sequence.

It does not build the tactile panel itself. Instead, it demonstrates how a short passage could be broken into individual words, so each word could be shown on the tactile panel in turn.

```python
# A short sample passage for the speculative tactile reading station
selected_passage = "God said Let there be light"

# Split the passage into individual words
words = selected_passage.split()

# Prepare the word-by-word sequence
reading_sequence = []

for position, word in enumerate(words, start=1):
    reading_sequence.append({
        "Step": position,
        "Word shown on tactile panel": word.upper(),
        "Audio confirmation": word.capitalize()
    })

reading_sequence
```

This table shows how the selected passage could be prepared for a tactile reading sequence. Each row represents one word that could appear on the pin-matrix panel.

The same sequence can also help with a practical design question: how wide would the tactile panel need to be for the longest word in the selected passage?

```python
# Find the longest word in the selected passage
longest_word = max(words, key=len)

# A simple design estimate:
# assume each letter needs 6 pin columns, plus 1 empty column between letters
pin_columns_per_letter = 6
spacing_between_letters = 1

estimated_panel_width = (
    len(longest_word) * pin_columns_per_letter
    + (len(longest_word) - 1) * spacing_between_letters
)

print(f"Longest word: {longest_word.upper()}")
print(f"Number of letters: {len(longest_word)}")
print(f"Estimated minimum pin columns needed: {estimated_panel_width}")
```

This code is not meant to solve the engineering problem of building a tactile display. It is included to show how the prototype’s interaction could be organised: a selected passage becomes a sequence of words, and the longest word helps estimate the space needed on the tactile panel.

## What this could suggest beyond one Bible

Although this prototype is designed around the raised-type Bible, it also suggests a wider question for cultural heritage access.

Many heritage objects depend on surface, shape, or texture, but cannot be freely touched because of conservation needs. A dynamic tactile panel could therefore be imagined as one way to interpret carved inscriptions, embossed decoration, seals, bookbindings, or object outlines without handling the original.

A simplified online version might also use haptic feedback, such as vibration on a trackpad or mobile device, to create a limited sense of rhythm or contact. This would not reproduce raised letters or material texture. However, it could still remind users that this book was designed for touch, not only for sight.

## Limits and ethical considerations

This prototype is speculative. It is not a working engineering model, and it does not claim to recreate the experience of nineteenth-century blind readers.

The dynamic pin panel would not feel the same as the original raised paper page. Its surface, scale, texture, and rhythm would all be different. The audio layer would also change the reading process by adding sound to an originally touch-based format.

The design also cannot be described as fully accessible without consultation and testing with blind and visually impaired users. In this portfolio, the station is presented as an exploratory concept for public interpretation, not as a finished accessibility solution.

Finally, the AI-generated images must be treated carefully. They are speculative visualisations, not archival evidence. Any final museum display would require accurate object information, permissions, and clear labelling.

## Conclusion: access beyond visibility

This raised-type New Testament shows that access is not only about seeing an object or reading information about it. For this book, access also depends on touch: the movement of fingers across raised letters, the pace of reading, and the physical contact with the page.

Digitisation can make the Bible easier to find, view, and share, but it can also make this touch-based experience harder to understand. The AI-assisted Dynamic Tactile Reading Station responds to that gap. It does not replace the original book or recreate nineteenth-century reading. Instead, it uses generative AI as a speculative design tool to imagine how touch might be brought back into cultural heritage access.

The project’s central point is therefore not that AI can solve accessibility. It is that AI, when guided by human interpretation and careful limits, can help us ask better questions about what access should include.
